```python
# ── Creación de listas ─────────────────────────────────────────────
numeros = [10, 20, 30, 40, 50]
mixta = [1, 'texto', 3.14, True] # tipos heterogéneos
vacia = []
de_rango = list(range(1, 11)) # [1, 2, 3, ..., 10]

# ── Acceso por índice ──────────────────────────────────────────────
print(numeros[0]) # 10 → primer elemento
print(numeros[-1]) # 50 → último elemento
print(numeros[-2]) # 40 → penúltimo

# ── Slicing [inicio : fin : paso] ───────────────────────────────
# 'fin' es EXCLUIDO — el rango es [inicio, fin)

print(numeros[1:4]) # [20, 30, 40] índices 1,2,3
print(numeros[:3]) # [10, 20, 30] desde el inicio hasta índice 2
print(numeros[2:]) # [30, 40, 50] desde índice 2 hasta el final
print(numeros[::2]) # [10, 30, 50] cada 2 elementos
print(numeros[::-1]) # [50, 40, 30, 20, 10] invertida

# ── Métodos principales ────────────────────────────────────────────
numeros.append(60) # agrega al final → O(1)
numeros.insert(2, 99) # inserta en posición 2 → O(n)
numeros.remove(99) # elimina primera ocurrencia
ultimo = numeros.pop() # extrae y devuelve el último elemento
print(len(numeros)) # longitud de la lista
```

## Ejercicio 2: Dado el diccionario datos = {'a':10, 'b':25, 'c':7, 'd':18}

### a) Crea una lista con las claves cuyo valor sea mayor a 12.

```python
# 1.- Definición del diccionario inicial
datos = {'a':10, 'b':25, 'c':7, 'd':18}

# 2.- Se crea una lista vacía para almacenar las claves que cumplan con la condición correspondiente
resultado = []

# 3.- Se usa ".items()" para obtener la clave con su valor correspondiente
for clave, valor in datos.items():

  # 4.- Se compara la variable "valor" con el número 12
  if valor > 12:
    # 5.- Si el valor es mayor a 12, se agrega la clave a la lista usando ".append()"
    resultado.append(clave)

print(resultado)
```

```python
# 1.- Definición del diccionario inicial
datos = {'a':10, 'b':25, 'c':7, 'd':18}

# 2.- Listado con las claves cuyo valor es mayor a 12
resultado = ['b', 'd']

# 3.- Se crea un diccionario nuevo usando las claves del resultado anterior y se asigna como valor el cuadrado del valor original.
# Se emplea las claves guardadas en la variable "resultado" y se eleva al cuadrado el valor correspondiente de cada clave.
cuadrados = {clave: datos[clave]**2 for clave in resultado}

# 4.- Se muestra el nuevo resultado
print(cuadrados)
```

## Ejercicio 3: Define una función 'resumen(lista)' que retorne un diccionario con: longitud, suma, promedio, mínimo y máximo de la lista.

```python
# 1.- Se define la función "resumen" que recibe una lista
def resumen(lista):
  # 2.- Se obtiene la cantidad de elementos que contiene la lista
  longitud = len(lista)
  # 3.- Se calcula la suma de todos los elementos de la lista
  suma = sum(lista)
  # 4.- Se calcula el promedio usando la suma y la longitud de la lista
  promedio = suma / longitud
  # 5.- Se obtiene el valor máximo y mínimo de la lista
  minimo = min(lista)
  maximo = max(lista)

  # 6.- Retorna un diccionario con los resultados obtenidos
  return{
      'longitud': longitud,
      'suma': suma,
      'promedio': promedio,
      'minimo': minimo,
      'maximo': maximo
  }

# 7.- Como en este ejercicio no se proporciona un listado de números específicos para este ejercicio, me planteé mi propia lista para la ejecución de esta parte
# del trabajo
n = [5, 8, 15, 17, 24, 31, 36, 41, 45, 52, 65]

# 8.- Se llama a la función "resumen()" utilizando una lista "n" como argumento, guardando el diccionario resultante en una variable
resultado = resumen(n)

# 9.- Se muestra el diccionario retornado por la función
print(resultado)
```

### a) Calcula la media, máximo y mínimo de cada columna.

```python
# 1.- Se importa la librería NumPy para acceder a sus funciones y tipos de datos.
import numpy as np
# 2.- Se fija la semilla del generador pseudoaleatorio en 7 para garantizar la reproductibilidad de los resultados.
np.random.seed(7)
# 3.- Se crea una matriz de 5x4 con enteros aleatorios entre 1 y 100 usando "np.random.randint()".
A = np.random.randint(1, 101, size=(5, 4))
# 4.- Se imprime la matriz inicial completa para visualizasr todos los valores generados.
print('Matriz inicial:\n', A)
# 5.- Se calcula la media de cada columna usando "axis=0", lo que indica que la operación se realiza de forma vertical a lo largo de la filas para obtener un promedio
# por cada columna.
print('\nMedia de cada columna:\n', A.mean(axis=0))
# 6.- Se calcula el valor máximo de cada columna usando "axis=0" para evaluar los elementos verticlamente por columna.
print('\nMáximo de cada columna:\n', A.max(axis=0))
# 7.- Se calcula el valor mínimo de cada columna usando "axis=0" para evaluar los elementos verticlamente por columna.
print('\nMínimo de cada columna:\n', A.min(axis=0))

# ¡¡¡¡IMPORTANTE!!!!
# axis = 0 (Eje de las filas/Vertical): Indica que la operación se realiza hacia abajo, recorriendo las filas por cada columna. Por lo tanto, el resultado final tendrá
# un valor por cada columna. Usar cuando realice cálculos estadísticos por columnas.
#
# axis = 1 (Eje de las columnas/Horizontal): Indica que la operación se realiza hacia los lados, recorriendo las columnas de cada fila. Por lo tanto, el resultado
# final tendrá un valor por cada fila. Usar cuando realice cálculos estadísticos por filas.
```

### b) Calcula la suma acumulada de la primera fila.


```python
# 1.- Se importa la librería NumPy para acceder a sus funciones y tipos de datos.
import numpy as np
# 2.- Se fija la semilla del generador pseudoaleatorio en 7 para garantizar la reproductibilidad de los resultados.
np.random.seed(7)
# 3.- Se crea una matriz de 5x4 con enteros aleatorios entre 1 y 100 usando "np.random.randint()".
A = np.random.randint(1, 101, size=(5, 4))
# 4.- Se imprime la matriz inicial completa para visualizasr todos los valores generados.
print('Matriz inicial:\n', A)
# 5.- Se selecciona solamente la primera fila de la matriz "(A[0])" y se calcula su suma acumulada usando la función "np.cumsum()", la cual va sumando secuencialmente
# cada elemento de izquierda a derecha.
print('\nSuma acumulada de la primera fila:\n', np.cumsum(A[0]))
```

### c) Normaliza la matriz para que cada columna tenga valores entre 0 y 1.
### Fórmula: (x - min_col) / (max_col - min_col)

```python
# 1.- Se importa la librería NumPy para acceder a sus funciones y tipos de datos.
import numpy as np
# 2.- Se fija la semilla del generador pseudoaleatorio en 7 para garantizar la reproductibilidad de los resultados.
np.random.seed(7)
# 3.- Se crea una matriz de 5x4 con enteros aleatorios entre 1 y 100 usando "np.random.randint()".
A = np.random.randint(1, 101, size=(5, 4))
# 4.- Se imprime la matriz inicial completa para visualizasr todos los valores generados.
print('Matriz inicial:\n', A)
# 5.- Se aplica la fórmula de normalización por columnas. A cada elemento de la matriz "A" se le resta el valor mínimo de su columna correspondiente y se divide entre
# la diferencia del valor máximo menos el mínimo de esa columna, logrando escalar todos los valores en un rango entre 0 y 1.
formula = ((A - A.min(axis=0)) / (A.max(axis=0) - A.min(axis=0)))
# 6.- Se imprime la matriz resultante con los valores normalizados
print('\nNormalización de la matriz:\n', formula)
```

## Ejercicio 3: Con el array de notas y nombres del ejemplo de argsort:

### a) ¿Cuántos estudiantes tienen nota mayor al promedio? (usa máscara booleana)

```python
# 1.- Se importa la librería NumPy para acceder a sus funciones y tipos de datos.
import numpy as np
# 2.- Se define un array unidimensional con los nombres de los estudiantes.
nombres = np.array(['Carlos','Ana','Diana','Pedro','Eva','Luis'])
# 3.- Se define un array unidimensional con las notas correspondientes a cada estudiante.
notas = np.array([75, 92, 88, 61, 95, 78])
# 4.- Se calcula el promedio de todas las notas utilizando "mean()".
promedio = notas.mean()
# 5.- Se crea una máscara booleana comparando cada nota con el promedio calculado.
mascara = notas > promedio
# 6.- Se utiliza el método "sum()" sobre la máscara booleana para contar cuantos valores "True" existen, para obtener el númmero total de estudiantes con nota mayor al
# promedio.
print('Número de estudiantes con una nota mayor al promedio:\n', mascara.sum())
```

## Ejercicio 4: Crea dos matrices 3×4 y concaténalas vertical y horizontalmente. Luego divide el resultado vertical en 3 partes iguales con vsplit.

```python
# 1.- Se importa la librería NumPy para acceder a sus funciones y tipos de datos.
import numpy as np
# 2.- Se crea la primera matriz "A" con un rango de 1 a 12 y se le da una forma de 3x4 usando "reshape()".
A = np.arange(1, 13).reshape(3,4)
# 3.- Se imprime la primera matriz creada para visualizar su contenido en la consola
print('Matriz 1:\n', A)
# 4.- Se crea la primera matriz "B" con un rango de 13 a 24 y se le da una forma de 3x4 usando "reshape()".
B = np.arange(13, 25).reshape(3,4)
# 5.- Se imprime la segunda matriz creada para visualizar su contenido en la consola
print('\nMatriz 2:\n', B)
# 6.- Se concatenan ambas matrices de forma vertical utilizando "np.vstack()" y se divide dicho resultado en tres partes iguales mediante la función "np.vsplit()".
vertical = np.vsplit(np.vstack((A, B)), 3)
# 7.- Se imprime de forma organizada y detallada cada una de las tres partes obtenidas tras el empleo de "vsplit()".
print('\nResultado de la concatenación vertical dividida en tres partes:')
print('Parte 1:\n', vertical[0])
print('\nParte 2:\n', vertical[1])
print('\nParte 3:\n', vertical[2])
# 8.- Se concatenan ambas matrices de forma horizontal, utilizando "np.hstack()".
horizontal = np.hstack((A, B))
# 9.- Se imprime el resultado final de la concatenación horizontal de las matrices.
print('\nConcatenación horizontal de las matrices:\n', horizontal)
```
